# Custom reward

`reward_fn` replaces the per-step reward. Milestone and terminal bonuses still come from the reward configuration unless those bonuses are set to zero there.

```python
from aRieL.rewards import RewardInfo

def t1_bonus(info: RewardInfo) -> float:
    if info.tier_completed == 1:
        return 1.0
    return 0.0
```

```python
from aRieL.validation import validate_reward

report = validate_reward(cfg="default", fn=t1_bonus)
print(report.summary_str())
report.raise_if_failed()
```

```python
from aRieL import ArielEnv, make_demo_targets

env = ArielEnv.from_preset(
    "demo", targets=make_demo_targets(8), reward_fn=t1_bonus
)
obs, info = env.reset(seed=0)
total_reward = 0.0
mask = info["action_mask"]
for _ in range(5):
    action = int(mask.nonzero()[0][0])
    obs, reward, terminated, truncated, info = env.step(action)
    mask = info["action_mask"]
    total_reward += float(reward)
    if terminated or truncated or not mask.any():
        break
print(f"steps {info['step_count']}  reward {total_reward:.3f}")
env.close()
```
